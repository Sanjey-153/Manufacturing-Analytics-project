```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Setup paths
PROJECT_ROOT = Path("..").resolve()
sys.path.append(str(PROJECT_ROOT))

from config import RAW_DATA_PATH, TARGET_COL, FAILURE_TYPE_COLS, NUMERICAL_COLS, CATEGORICAL_COLS
print(f"Loading dataset from: {RAW_DATA_PATH}")
df = pd.read_csv(RAW_DATA_PATH)
df.head()
```

## 1. Dataset Dimensions and Data Types

```python
print(f"Total Observations (Rows): {df.shape[0]:,}")
print(f"Total Attributes (Columns): {df.shape[1]}")
df.info()
```

## 2. Data Integrity, Missing Values, and Duplicates

```python
print("Missing values per column:")
print(df.isnull().sum())
print(f"\nExact duplicate rows: {df.duplicated().sum()}")
```

## 3. Sensor Numerical Descriptive Statistics

```python
stats_df = df[NUMERICAL_COLS].describe().T
stats_df["median"] = df[NUMERICAL_COLS].median()
stats_df["skewness"] = df[NUMERICAL_COLS].skew()
stats_df
```

## 4. Target Class Distribution & Imbalance

```python
target_counts = df[TARGET_COL].value_counts()
target_pct = df[TARGET_COL].value_counts(normalize=True) * 100
print("Machine Failure Counts:")
print(target_counts)
print("\nMachine Failure Percentages:")
print(target_pct.round(2))
print(f"\nClass Imbalance Ratio: {target_counts[0]/target_counts[1]:.2f}:1")
```

## 5. Specific Failure Modes Breakdown

```python
for f in FAILURE_TYPE_COLS:
    print(f"{f}: {df[f].sum()} occurrences ({df[f].mean()*100:.2f}% of dataset)")
```
